### Questão 2
Uma pesquisa deseja investigar se a proporção de profissionais que utilizam Python é diferente entre aqueles que trabalham remotamente e aqueles que não trabalham remotamente.

Utilize as variáveis uses_python e is_remote.

Realize um teste de hipótese para determinar se existe diferença estatisticamente significativa entre as duas proporções, com confiança de 95%.

Orientações: Apresente as proporções observadas em cada grupo e interprete o resultado do teste.

H₀: p_remoto = p_não remoto \
H₁: p_remoto ≠ p_não remoto

```python
d = df.dropna(subset=['is_remote','uses_python'])
r = d[d.is_remote==True]
nr = d[d.is_remote==False]
x1,n1 = r.uses_python.sum(),len(r)
x2,n2 = nr.uses_python.sum(),len(nr)
```

```python
print(f"Remotos: {x1/n1:.4f}")
print(f"Não remotos: {x2/n2:.4f}")
```

```python
proportions_ztest([x1, x2],[n1, n2])
```

Conclusão: \
Temos pValue < 0.05. Assim, rejeitamos a hipótese nula. Existe diferença estatisticamente significativa entre as proporções. A proporção de profissionais que utilizam Python é maior entre os classificados como não remotos.

Realize um teste de hipótese para determinar se há evidência estatística de que imóveis com ar-condicionado central possuem preço médio superior.

Para essa análise, considere apenas os imóveis com valores de OverallQual iguais a 5, 6, 7 ou 8.

```python
resultado = tukey_hsd(*grupos)
print(resultado)
```

### Questão 5
Uma empresa fabrica componentes eletrônicos cujo tempo de funcionamento apresenta distribuição aproximadamente normal. O fabricante afirma que a variância populacional do tempo de funcionamento é de 100 horas². Uma amostra aleatória de 25 componentes apresentou desvio padrão amostral de 13 horas. Realize um teste de hipótese para investigar se a variância populacional é diferente do valor informado pelo fabricante.

H₀: σ² = 100 \
H₁: σ² ≠ 100

```python
n, s, sigma2 , alpha= 25, 13, 100, .05
gl = n-1
chi = (gl*s*s)/sigma2
lo, hi = chi2.ppf([alpha/2,1-alpha/2], gl)
print('Qui-quadrado:', chi, 'limites:', lo, hi)
print('Rejeitar H0' if chi<lo or chi>hi else 'Não rejeitar H0')
```

Conclusão: \
Como 40,56 é maior que 39,3641, rejeitamos H₀. Há evidências estatísticas de que a variância populacional é diferente de 100 horas².

### Questão 6
Uma empresa de tecnologia desenvolveu um novo método de treinamento para melhorar a produtividade de programadores. Historicamente, o tempo médio necessário para concluir uma determinada tarefa de programação é de 120 minutos, com desvio padrão populacional conhecido de 24 minutos. Após o treinamento, uma amostra aleatória de 64 programadores apresentou tempo médio de 113 minutos. Considere que os tempos são independentes e que o desvio padrão populacional permanece igual a 24 minutos. Realize um teste de hipótese com confiança de 95% para investigar se o treinamento reduziu o tempo médio necessário para concluir a tarefa.

Orientações: Apresente as hipóteses, a estatística calculada, o p-valor e a conclusão.

H₀: μ ≥ 120 \
H₁: μ < 120

```python
mu0 = 120
sigma = 24
x_bar = 113
n = 64
alpha = 0.05

# Estatística Z
z = (x_bar - mu0) / (sigma / np.sqrt(n))

# Teste unilateral à esquerda
p_value = norm.cdf(z)

print(f"Estatística Z: {z:.4f}")
print(f"p-valor: {p_value:.6f}")

if p_value < alpha:
    print("Rejeitar H0")
else:
    print("Não rejeitar H0")
```

Conclusão: \
Rejeitamos H₀. Há evidências estatísticas de que o tempo médio necessário para concluir a tarefa é inferior a 120 minutos após o treinamento.

### Questão 7
Uma equipe de pesquisadores deseja investigar se a variabilidade da massa corporal dos pinguins da espécie Adelie é diferente entre machos e fêmeas. Utilizando o dataset penguins.csv, considere as seguintes variáveis:
- species: espécie do pinguim.
- sex: sexo do pinguim (male ou female).
- body_mass_g: massa corporal em gramas.
Realize um teste de hipótese para verificar se existe diferença estatisticamente significativa entre as variâncias populacionais da massa corporal de machos e fêmeas da espécie Adelie.

Orientações: Considere apenas registros completos nas variáveis de interesse (use dropna()). Apresente as hipóteses, a estatística de teste e a conclusão.

H₀: σ²_machos = σ²_fêmeas \
H₁: σ²_machos ≠ σ²_fêmeas

```python
adelie = df[df["species"] == "Adelie"]

machos = adelie.loc[
    adelie["sex"] == "male", "body_mass_g"
].dropna()

femeas = adelie.loc[
    adelie["sex"] == "female", "body_mass_g"
].dropna()

print("Machos:", len(machos))
print("Fêmeas:", len(femeas))
```

```python
alpha = 0.05

var_m = machos.var(ddof=1)
var_f = femeas.var(ddof=1)

n_m = len(machos)
n_f = len(femeas)

F_cal = var_m / var_f
print(f'F_cal: {F_cal}')

df1 = n_m - 1
df2 = n_f - 1

print (f.isf(0.025, df1, df2))
print (f.isf(0.975, df1, df2))
```

Conclusão: 
Temos F_cal = 1.66 fora do intervalo 0.62 a 1.59. Assim, rejeitamos a hipótese nula. Há evidências estatísticas suficientes para afirmar que as variâncias populacionais da massa corporal dos pinguins Adelie machos e fêmeas são diferentes. A variância amostral dos machos é aproximadamente 65,8% maior que a das fêmeas.
