```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
```

Questão 1 \
O peso médio de 500 estudantes do sexo masculino de uma universidade é 75,50 kg e o desvio padrão 7,50 kg. Admitindo-se que os pesos estão normalmente distribuídos e que as medidas foram realizadas com uma balança que registrou valores na escala de centésimos de quilogramas, determine quantos estudantes pesam: \
a) entre 60 e 77,5 kg \
b) mais do que 92,5 kg \
c) menos do que 64 kg \
d) 64 kg \
e) 64 kg ou menos

```python
mu_1 = 75.50
sigma_1 = 7.50
total_estudantes = 500

print("--- Questão 1: Pesos ---")

# a) Entre 60 e 77,5 kg
prob_a = norm.cdf(77.50, loc=mu_1, scale=sigma_1) - norm.cdf(60, loc=mu_1, scale=sigma_1)
print(f"a) Probabilidade entre 60 e 77,5 kg: {prob_a:.4f} -> {int(prob_a * total_estudantes)} estudantes")
```

```python
# b) Mais do que 92,5 kg
# Como queremos a área à direita, subtraímos a área acumulada de 1.
prob_b = 1 - norm.cdf(92.5, loc=mu_1, scale=sigma_1)
print(f"b) Probabilidade mais de 92,5 kg: {prob_b:.4f} -> {int(prob_b * total_estudantes)} estudantes")
```

```python
# c) Menos do que 64 kg
prob_c = norm.cdf(64, loc=mu_1, scale=sigma_1)
print(f"c) Probabilidade menos de 64 kg: {prob_c:.4f} -> {int(prob_c * total_estudantes)} estudantes")
```

```python
# d) Exatamente 64 kg
# Distribuições contínuas têm probabilidade zero para pontos exatos.
prob_d = 0.0
print(f"d) Probabilidade exatamente 64 kg: {prob_d:.4f} -> 0 estudantes")
```

```python
# e) 64 kg ou menos
print(f"e) Probabilidade 64 kg ou menos: {prob_c:.4f} -> {int(prob_c * total_estudantes)} estudantes\n")
```

Questão 2 \
Aparelhos de TVs são substituídos normalmente com uma média de 8,2 anos e desvio padrão de 1,1 ano.
a) Determine a probabilidade de que uma TV deva ser substituída em menos de 5,0 anos. \
b) Se você deseja estabelecer uma garantia de modo que apenas 1% dos aparelhos seja substituídos antes da expiração da garantia, qual deve ser o prazo ? Dica: use a função ppf

```python
mu_2 = 8.2
sigma_2 = 1.1

print("--- Questão 2: Substituição de TVs ---")

# a) Probabilidade de que uma TV seja substituída em menos de 5,0 anos
prob_2a = norm.cdf(5.0, loc=mu_2, scale=sigma_2)
print(f"a) Probabilidade em menos de 5,0 anos: {prob_2a * 100:.4f}%")
```

```python
# b) Prazo para que apenas 1% (0.01) dos aparelhos sejam substituídos antes da expiração
# Usamos norm.ppf (Percent Point Function), que é a inversa da CDF
prazo_garantia = norm.ppf(0.01, loc=mu_2, scale=sigma_2)
print(f"b) Prazo de garantia para 1%: {prazo_garantia:.2f} anos")
```

Questão 3 \
A média dos diâmetros internos de uma amostra de 200 arruelas produzidas por certa máquina é 0,502” e o desvio padrão 0,005”. A finalidade para a qual essas arruelas são produzidas permite a tolerância máxima de 0,496 a 0,508 polegadas de diâmetro. Determine a porcentagem de arruelas defeituosas produzidas, admitindo-se que os diâmetros são normalmente distribuídos.

```python
mu_3 = 0.502
sigma_3 = 0.005
total_arruelas = 200

print("\n--- Questão 3: Arruelas Defeituosas ---")

# Probabilidade de uma arruela estar dentro dos limites de tolerância (entre 0.496 e 0.508)
prob_dentro = norm.cdf(0.508, loc=mu_3, scale=sigma_3) - norm.cdf(0.496, loc=mu_3, scale=sigma_3)

# A porcentagem de arruelas defeituosas é o complemento daquelas que estão aceitáveis
prob_defeituosas = 1 - prob_dentro
print(f"Porcentagem de arruelas defeituosas: {prob_defeituosas * 100:.2f}%")
print(f"Quantidade aproximada de defeituosas: {int(prob_defeituosas * total_arruelas)} arruelas")
```

Questão 4 \
Suponha que as temperaturas do corpo humano sejam normalmente distribuídas, com média 36,8°C e desvio padrão 0,35°C. \
a) Um hospital usa 38,1°C como a menor temperatura considerada como febre. Qual é porcentagem de pessoas normais e saudáveis que seriam consideradas febris? Tal porcentagem sugere que o corte em 38,1°C seja apropriado? \
b) Os médicos desejam selecionar uma temperatura mínima para requerimento de exames adicionais. Qual deve ser essa
temperatura, se eles desejam que somente 5% das pessoas saudáveis tenham temperatura superior ? (Tal resultado é chamado falso positivo. Teste positivo, mas a pessoa não está doente).

```python
mu_4 = 36.8
sigma_4 = 0.35

print("\n--- Questão 4: Temperaturas do Corpo Humano ---")

# a) Menor temperatura considerada febre = 38,1°C. Qual a porcentagem de pessoas saudáveis consideradas febris?
# Queremos a probabilidade de ter uma temperatura SUPERIOR a 38,1°C
prob_febre = norm.sf(38.1, loc=mu_4, scale=sigma_4)
print(f"a) Porcentagem de falsos positivos (febres em saudáveis): {prob_febre * 100:.4f}%")
```

```python
# b) Temperatura mínima para exames adicionais se desejam que apenas 5% (0.05) das pessoas saudáveis tenham temperatura superior
# Como queremos que 5% fiquem acima, procuramos o percentil de 95% (1 - 0.05)
temp_limite = norm.ppf(0.95, loc=mu_4, scale=sigma_4)
print(f"b) Temperatura mínima para exames (5% superior): {temp_limite:.2f}°C")
```
