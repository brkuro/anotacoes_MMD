## Criação de histogramas com Seaborn
Nesta aula analisaremos o índice de percepção de corrupção no setor público nos diferentes países do globo.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('CPI_2014.csv', encoding='latin1')
```

```python
df.head()
```

```python
sns.displot(df['CPI_2014'], kde = False)
sns.set_style('darkgrid')
plt.title('Indice de corrupção (Ano 2014)')
plt.xlabel('Nivel de Corrupção')
plt.axvline(60, color = 'k', linestyle = '--')
plt.show()
```

Exercício \
Criar um histograma apenas com países da Europa.

```python
df2 = df[df['Continent'] == 'Europe']
df2
```

```python
sns.displot(df2['CPI_2014'],bins = 15, kde = False)
sns.set_style('darkgrid')
plt.title('Indice de corrupção (Ano 2014)')
plt.xlabel('Nivel de Corrupção')
plt.axvline(60, color = 'k', linestyle = '--')
plt.show()
```
