# Aula 24 - Introdução à Álgebra Linear

Nesta aula, vamos explorar os conceitos básicos de vetores, matrizes e sistemas lineares.

## 1) Vetores
Um vetor é uma estrutura com magnitude e direção. Em Python, usamos arrays do NumPy.

```python
import numpy as np

u = np.array([2, -1, 3])
v = np.array([1, 4, -2])

print('u =', u)
print('v =', v)
print('u + v =', u + v)
print('2*u =', 2 * u)
print('produto interno u · v =', np.dot(u, v))
```

## 2) Matrizes
Uma matriz é uma tabela de números organizada em linhas e colunas.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('A =\n', A)
print('B =\n', B)
print('A + B =\n', A + B)
print('A @ B =\n', A @ B)
```

## 3) Determinante e matriz inversa
O determinante indica se uma matriz pode ser invertida.

```python
M = np.array([[2, 1], [4, 3]])
detM = np.linalg.det(M)
invM = np.linalg.inv(M)

print('M =\n', M)
print('det(M) =', detM)
print('M^-1 =\n', invM)
print('M * M^-1 =\n', M @ invM)
```

## 4) Sistema linear
Vamos resolver o sistema:
2x + y = 5
x + 3y = 8

```python
A = np.array([[2, 1], [1, 3]], dtype=float)
b = np.array([5, 8], dtype=float)

x = np.linalg.solve(A, b)

print('Solução do sistema:', x)
print('Verificação: A@x =', A @ x)
```

## 5) Aplicação simples
Suponha que uma loja vende três tipos de produtos em diferentes quantidades. A receita total pode ser calculada com um produto escalar.

```python
precos = np.array([120, 80, 150])
quantidades = np.array([3, 5, 2])

faturamento = np.dot(precos, quantidades)
print('Preço dos produtos:', precos)
print('Quantidades vendidas:', quantidades)
print('Faturamento total = R$', faturamento)
```

```python
F = np.array([[6, 3, 10], [4, 8, 5]])
P = np.array([[0.10, 0.15], [0.40, 0.30], [0.10, 0.20]])
print("Quantidade de Frutas:")
print(F)
print()
print("Preços")
print(P)
print()
print("Custo total: F@P:")
R = F@P
print(R)
print()
if R[0,0] < R[0,1]:
    print(f'Quitanda A tem o melhor preço João {R[0,0]}')
else:
    print(f'Quitanda B tem o melhor preço João {R[0,1]}')
if R[0,1] < R[1,1]:
    print(f'Quitanda A tem o melhor preço Maria {R[0,1]}')
else:
    print(f'Quitanda B tem o melhor preço Maria{R[1,1]}')
```
